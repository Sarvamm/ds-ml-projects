```python
import pandas as pd
import random as rd

pd.Series(
    [rd.randint(1,6) for i in range(10_000)]
).value_counts().sort_index().plot(kind='bar', rot=0)
```

```python
import plotly.express as px
import pandas as pd
import random as rd

def toss_n_times(n):
    return [rd.choice(('Heads', 'Tails')) for i in range(n)]

def experiment( outcome='Heads', n = 100, times=10_000 ):
    return px.histogram(
        data_frame = pd.Series( [toss_n_times(n).count(outcome) for i in range(times)] ),
        title='Binomial Distribution of Coin Tosses',
        labels={'value':f'Number of {outcome} in {n} tosses'},
        width=800, height=400
    )

experiment()
```

```python
import pandas as pd
import random as rd

pd.Series(
    [rd.randint(0,4) for i in range(10)] + [rd.choice((2,3)) for i in range(5)]
).hist(grid=False, figsize=(6,4))
```

```python
import numpy as np
import plotly.express as px
from scipy.stats import norm

#range
x = np.linspace(12, 38, 100)

# loc = mean, scale = standard deviation
y = norm.pdf(x, loc=25, scale=3.5)

px.area(x=x, y=y,
        title='Probability distribution of heights of maple trees',
        labels={
            'y':'Probability Density',
            'x':'Height (meters)'
        },
         width=800, height=450, )
```
