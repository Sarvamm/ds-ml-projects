```python
import pandas as pd
import numpy as np
import plotly.express as px
from plotly.subplots import make_subplots

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
```

```python
from sklearn.datasets import load_wine
```

```python
data = load_wine()    
```

```python
y = data.target # pyright: ignore[reportAttributeAccessIssue]
X = data.data # pyright: ignore[reportAttributeAccessIssue]
```

```python
X_train, x_, y_train, y_ = train_test_split(X, y, test_size=0.3, random_state=42)
X_test, X_val, y_test, y_val = train_test_split(x_, y_, test_size=0.5, random_state=42)
```

```python
poly = PolynomialFeatures(degree=2)
X_train_mapped = poly.fit_transform(X_train)
X_test_mapped = poly.transform(X_test)
X_val_mapped = poly.transform(X_val)
```

```python
scaler = StandardScaler()
X_train_norm = scaler.fit_transform(X_train_mapped)
X_test_norm = scaler.transform(X_test_mapped)
X_val_norm = scaler.transform(X_val_mapped)
```

```python
dt_model = DecisionTreeClassifier()
dt_model.fit(X_train_norm, y_train)

dt_model_pred = dt_model.predict(X_val)
print(accuracy_score(y_val, dt_model_pred))
```

```python
simple_model = LogisticRegression(multi_class='multinomial', solver='lbfgs')
simple_model.fit(X_train_norm, y_train)

simple_model_pred = simple_model.predict(X_val)
print(accuracy_score(y_val, simple_model_pred))
```
