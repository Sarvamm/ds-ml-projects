```python
import numpy as np
import plotly.express as px
from typing import Callable
import random
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.datasets import mnist
from tensorflow.keras import regularizers
from tensorflow.keras.layers import Dropout
```

```python
df = mnist.load_data()
```

```python
X_train, y_train = df[0]
X_test, y_test = df[1]  
```

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
def show_random_sample():
    random_num = random.randint(0, X_train.shape[0])
    return px.imshow(X_train[random_num], title=f'Label: {y_train[random_num]}' )
```

```python
show_random_sample()
```

```python
X_train_flattened = X_train.reshape(X_train.shape[0], -1)
X_test_flattened = X_test.reshape(X_test.shape[0], -1)
```

```python
X_train_flattened.shape
```

```python
mnist_model = Sequential(
    [
        Input(shape=(784,)),
        
        Dropout(0.2),

        Dense(128, activation='relu'), 
        
        Dense(32, activation='relu'),
        
        Dense(16, activation='relu'),

        Dense(10, activation='softmax')
    ],
    name = "MNIST_Neural_Network"
)

mnist_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
mnist_model.summary()
```

```python
history = mnist_model.fit(X_train_flattened, y_train,
                          epochs=40,
                          verbose=2,
                          batch_size=64,
                          validation_data =(X_test_flattened, y_test))
```

```python
from plotly.subplots import make_subplots
```

```python
def plot_results(start_epoch=40):
    

    acc= history.history['accuracy'][start_epoch:]
    loss = history.history['loss'][start_epoch:]
    val_acc = history.history['val_accuracy'][start_epoch:]
    val_loss = history.history['val_loss'][start_epoch:]

    fig = make_subplots(2,2, subplot_titles=("Accuracy", "Loss", "Val Accuracy", "Val Loss"))

    fig.add_trace(px.line(x=history.epoch[start_epoch:], y=acc).data[0], 1, 1)
    fig.add_trace(px.line(x=history.epoch[start_epoch:], y=loss).data[0], 1, 2)
    fig.add_trace(px.line(x=history.epoch[start_epoch:], y=val_acc).data[0], 2, 1)
    fig.add_trace(px.line(x=history.epoch[start_epoch:], y=val_loss).data[0], 2, 2)

    fig.show()

plot_results(20)
```

# Neural Network with Numpy

```python
# --------------------------- ACTIVATION FUNCTIONS --------------------------- #

#Sigmoid
def sigmoid( z: float | int ) -> float:
    return 1 /( 1 + np.exp(-z))

#ReLU
def relu( z: float | int ) -> float:
    return np.maximum( 0 , z)     

#Softmax
def softmax( z: np.ndarray ) -> np.ndarray:
    return np.exp(z) / np.sum(np.exp(z))   
   
# ----------------------------- PLOTTING FUNCTION ---------------------------- #
def plot_fn( fn: Callable ):
    x = np.linspace(-10,10)
    y = fn(x)

    return px.line(x=x, y=y)

# ---------------------------- NORMALISE FUNCTION ---------------------------- #
def normalise( X: np.ndarray ) -> np.ndarray:
    return (X - np.mean(X)) / np.std(X)

```

```python
def dense(A_in: np.ndarray, #(ndarray (m,n))
          W: np.ndarray, # (ndarray (n,j))
          B: np.ndarray, # (ndarray (1,j))
          g: Callable = relu) -> np.ndarray: 
      
      """
      Computes dense layer
      Args:
      A_in (ndarray (m,n)) : Data, m examples, n features each
      W    (ndarray (n,j)) : Weight matrix, n features per unit, j units
      b    (ndarray (1,j)) : bias vector, j units  
      g    activation function (e.g. sigmoid, relu..)
      Returns
      A_out (tf.Tensor or ndarray (m,j)) : m examples, j units
      """

             
      return( g( A_in @ W + B ) )
```

```python
#save weight from mnist model
W1 = mnist_model.layers[1].get_weights()[0]
b1 = mnist_model.layers[1].get_weights()[1]
W2 = mnist_model.layers[2].get_weights()[0]
b2 = mnist_model.layers[2].get_weights()[1]
W3 = mnist_model.layers[3].get_weights()[0]
b3 = mnist_model.layers[3].get_weights()[1]
W4 = mnist_model.layers[4].get_weights()[0]
b4 = mnist_model.layers[4].get_weights()[1]
```

```python
#shapes
print(W1.shape)
print(b1.shape)
print(W2.shape)
print(b2.shape)
print(W3.shape)
print(b3.shape)
print(W4.shape)
print(b4.shape)
```

```python
def sequential(X, W1, b1, W2, b2, W3, b3, W4, b4):
    a1 = dense(X,  W1, b1, relu)
    a2 = dense(a1, W2, b2, relu)
    a3 = dense(a2, W3, b3, relu)
    a4 = dense(a3, W4, b4, softmax)
    return(a4)
```

```python

def predict(X):
    
    return( sequential( X, W1, b1, W2, b2, W3, b3, W4, b4    ) )
```

```python
def test():
    random_choice = random.randint(0, len(X_test) - 1)
    test_input = X_test[random_choice].reshape(1,-1)
    correct_answer = y_test[random_choice]
    prediction = predict(test_input)
    
    fig = make_subplots(1,2)
    fig.add_trace(px.histogram(x=range(10), y=prediction[0], nbins=20).data[0], 1, 1)
    fig.add_trace(px.imshow( X_test[random_choice][::-1] , title=f"True label: {correct_answer}" ).data[0], 1, 2)
    fig.update_layout(height=500, width=1000, title_text= f'       {prediction[0].max() * 100: .4f} % sure that this is a  {np.argmax(prediction[0])}\
                                            True label: {correct_answer}')
    fig.show()
    
```

```python
test()

```
